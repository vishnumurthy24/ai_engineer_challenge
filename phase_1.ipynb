Problem statement

Create a Python program that stores configuration settings for multiple AI models in structured objects.

Your program should support:

Model name

Provider

Temperature

Maximum tokens

System prompt

Printing configuration details

```python
from dataclasses import dataclass

@dataclass
class AIconfig:
    model_name:str
    provider:str
    temperature:float
    maximum_token:int
    system_prompt:str

chat_config=AIconfig(
    model_name="gpt-4o-mini",
    provider="openai",
    temperature=0.7,
    maximum_token=2000,
    system_prompt="You are a helpful assistant."
)
code_config=AIconfig(
    
    model_name="gpt-4o-mini",
    provider="openai",
    temperature=0.2,
    maximum_token=2000,
    system_prompt="You are a helpful assistant that writes code."
)

print(chat_config)
print(code_config)
```

```python
from functools import wraps

def log_call(func):
    @wraps(func)
    def wrapped(*args, **kwargs):
        print("function started")
        result = func(*args,**kwargs)
        
        print("function ended")
        return result
    return wrapped
```

Task 3 — Add timing

Create a timer decorator that prints how many seconds a function takes. Use time.perf_counter()

```python
import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapped(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start} seconds to execute.")
        return result
    return wrapped
```

```python
@timer
def heavy_calculation(n):
    # Simulating a heavy computation loop
    return sum(i * i for i in range(n))

# Testing the decorator
result = heavy_calculation(1_000_000)
```

Task 1 — Number generator

```python
def num_gen(n):
    for i in range(1,n):
        yield i
```

```python
for num in num_gen(10):
    print(num)
```

Task 2 — Even number generator

```python
def get_even(n):
    for i in range(1,n):
        if i % 2 == 0:
            yield i 
```

```python
for even in get_even(54):
    print(even)
```

Task 2 — Create your own context manager

```python
class aimsg:
    def __enter__(self):
        print("the process has started")
        return self

    def __exit__(self,exc_type,exc_value,traceback):
        print("the process has ended")
        return self

with aimsg():
    print("the process is running.......")
```

Task 3 — Bonus
Use @contextmanager

```python
from contextlib import contextmanager

@contextmanager
def Ai_process():
    print("task is assigned")
    yield
    print("task is completed")

with Ai_process():
    print("the task is running.......")
```
