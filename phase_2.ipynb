```python
import numpy as np

sample=np.array([10,30,4,20,78,89,65])
```

```python
print(sample.ndim)
print(sample.shape)
print(sample.size)
print(sample.dtype)
```

```python
marks = np.array([75, 82, 91, 68, 88])

print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
```

```python
students = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [95, 90, 92]
])
```

```python
print(students.shape)
print(students[0,:])
print(students[1,1])
print(students[:,0])
print(np.mean(students))
```
