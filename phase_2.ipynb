```python
marks = np.array([75, 82, 91, 68, 88])

print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
```

```python
students = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [95, 90, 92]
])
```

```python
print(students.shape)
print(students[0,:])
print(students[1,1])
print(students[:,0])
print(np.mean(students))
```

***condition***

Numbers less than 50



Numbers between 30 and 80


Numbers greater than or equal to 70

```python
numbers = np.array([10, 25, 40, 55, 70, 85, 100])
```

```python

print(numbers[numbers > 50])
print(numbers[numbers < 50])
print(numbers[(numbers>30)&(numbers<80)])
print(numbers[numbers>=70])
```
